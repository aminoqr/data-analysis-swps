## 1. Selected arithmetic operations on a NumPy array

```python
import numpy as np
```

```python
# Functions taken from https://tobiasraabe.github.io/post/numpy-views-vs-copies/
# These functions make it easier to determine whether two NumPy arrays have a common base
# which means they share memory (or not)

def get_data_base(x):
    base = x
    while isinstance(base.base, np.ndarray):
        base = base.base
    return base

def arrays_share_data(x, y):
    return get_data_base(x) is get_data_base(y)
```

```python
arr = np.arange(1,101).reshape((10,10))
arr
```

```python
# sum of all values in the array
print(arr.sum())
# lub
print(np.sum(arr))
```

```python
# sum in each column (row indexing)
print(arr.sum(axis=0))
# and each rows (column indexing)
print(arr.sum(axis=1))
```

```python
# cumulative sum
print(arr.cumsum())
```

```python
# most of the functions can be run for a given axis
print(arr.cumsum(axis=0))
print(arr.cumsum(axis=1))
```

```python
# For the purposes of a few examples, we will use a copy of the original array
arr_nan = arr.copy()
```

```python
# Filtering an array returns a Boolean array—a map—of values that satisfy and do not satisfy the condition
divide_by_five_map = (arr_nan % 5 == 0)
divide_by_five_map
```

```python
# We'll also check from time to time whether we're working on a copy or a view
# Here we're dealing with what's known as “fancy slicing”—so it's a copy
arrays_share_data(arr_nan, divide_by_five_map)
```

```python
# We can attach this panel to another panel (make sure the dimensions match!)
divide_by_five = arr_nan[divide_by_five_map]
divide_by_five
```

```python
arrays_share_data(divide_by_five, arr_nan)
```

```python
# We convert the values associated with the map indices to values such as “nan” (not a number)
arr_nan[divide_by_five_map] = np.nan
```

```python
arr_nan
```

```python
arr_nan.dtype
```

```python
# It turns out that we can't do this for types like `int`
# Maybe for types like `float`
# After a failed insertion operation, we should check whether the array contains valid data
# If not, the copying operation should be performed again
arr_nan = arr_nan.astype(np.float64)
arr_nan
```

```python
arr_nan[divide_by_five_map] = np.nan
```

```python
# Let's perform the sum operation again
print(arr_nan.sum())
```

```python
print(arr_nan.sum(axis=0))
print(arr_nan.sum(axis=1))
```

```python
# If we want to treat nan values in a specific way, e.g., as 0 (zero), we have
# dedicated functions at our disposal—in this case, the sum function
print(np.nansum(arr_nan))
print(np.nansum(arr_nan, axis=0))
print(np.nansum(arr_nan, axis=1))
```

```python
# Multiplying array elements
# Not working? Why 0? This is due to an overflow of values for types supported by NumPy.
# Python's built-in int type can represent extremely large numbers (limited only by RAM), but  numpy
# is based on types known from the C language. Value overflow for types like int64 or float64 during multiplication can
# occur very quickly. In this case, 0 is returned without any warning—keep this in mind.
print(arr.prod())
print(arr.prod(axis=0))
```

```python
# NumPy has a wide range of built-in functions for arithmetic, trigonometry, linear algebra, and more.
# We’ll cover linear algebra a bit later. Below are a few examples of other common operations.
print(arr.max(), arr.max(axis=0))
```

```python
# We can also use operations of this type to filter an array
arr[(arr == arr.max(axis=0))]
```

```python
# Here, it’s also worth introducing a function like `where`, which, instead of returning values that meet a condition,
# can return the indices where those conditions are met. Below, we’ll see two arrays, since the array
# we’re filtering is two-dimensional. The first dimension is the row, and the second is the column.
# The `where` function takes up to three arguments: condition, x, y. If only the condition is provided, it returns the indices of the elements
# that satisfy that condition.
```

```python
np.where(arr == arr.max(axis=0))
```

```python
# What is the equivalent of the following?
np.asarray(arr == arr.max(axis=0)).nonzero()
```

```python
# but we can also use it this way
# condition, value if the condition is met, value if the condition is not met
np.where(arr == arr.max(axis=0), arr, np.nan)
```


6. Insert the following into the `no_even_small_one` array using the `np.where` function:

   * the value 1 for each `np.nan` value (see the built-in `np.isnan()` function)
   * even numbers, which we previously replaced with `nan` values (now 1 if the first step is completed), so we restore the array to its original state from the `a_small_one` array—for the condition, insert the value from `a_small_one` or `no_even_small_one`




```python
import numpy as np

a_small_one = np.arange(1, 65).reshape(8, 8)
no_even_small_one = a_small_one.astype(float)
for r in range(8):
    for c in range(8):
        if a_small_one[r, c] % 2 == 0:
            no_even_small_one[r, c] = np.nan

# Task 6: put 1 where NaN, then restore real evens from a_small_one
step = np.where(np.isnan(no_even_small_one), 1.0, no_even_small_one)
mask_even = a_small_one % 2 == 0
restored = np.where(mask_even, a_small_one.astype(float), step)
print(restored)
print("Matches original:", np.allclose(restored, a_small_one.astype(float)))

```


8. For the `normal_100` array, calculate:
   * the mean value in each row and column (separately),
   * calculate the average absolute difference for the values in the previous point relative to the value 0 (this is the expected value given our assumptions about the normal distribution) — separately for the column averages and the row averages; keep in mind that operations on a NumPy array are performed in broadcast mode.




```python
import numpy as np

rng = np.random.default_rng(seed=0)
normal_100 = rng.standard_normal((100, 100))

# Mean of each row
row_means = []
for i in range(100):
    row_means.append(normal_100[i, :].mean())

# Mean of each column
col_means = []
for j in range(100):
    col_means.append(normal_100[:, j].mean())

row_means = np.array(row_means)
col_means = np.array(col_means)

# Average absolute distance of those means from 0
sum_row = 0.0
for v in row_means:
    sum_row += abs(v - 0.0)
avg_abs_row = sum_row / len(row_means)

sum_col = 0.0
for v in col_means:
    sum_col += abs(v - 0.0)
avg_abs_col = sum_col / len(col_means)

print("Avg |row means| from 0:", avg_abs_row)
print("Avg |col means| from 0:", avg_abs_col)

```


9. Create a 1000x1000 array by filling it with random values from a standard normal distribution, and then, using examples from previous labs (the `timeit` module), compare the execution times for this array:
    * writing to a text file vs. writing to a file in pickle format vs. writing to a compressed file (the `savez_compressed` function in the NumPy library documentation)
    * reading arrays back from the three files mentioned above.



```python
import numpy as np
import pickle
import os
import tempfile
from timeit import timeit

# Task 9: time saving/loading big array three ways
rng = np.random.default_rng(1)
big = rng.standard_normal((1000, 1000))
folder = tempfile.mkdtemp()
path_txt = os.path.join(folder, "arr.txt")
path_pkl = os.path.join(folder, "arr.pkl")
path_npz = os.path.join(folder, "arr.npz")


def write_txt():
    np.savetxt(path_txt, big)


def read_txt():
    return np.loadtxt(path_txt)


def write_pkl():
    with open(path_pkl, "wb") as f:
        pickle.dump(big, f)


def read_pkl():
    with open(path_pkl, "rb") as f:
        return pickle.load(f)


def write_npz():
    np.savez_compressed(path_npz, data=big)


def read_npz():
    loaded = np.load(path_npz)
    return loaded["data"]


runs = 2
print("write txt:", timeit(write_txt, number=runs))
print("write pkl:", timeit(write_pkl, number=runs))
print("write npz:", timeit(write_npz, number=runs))
print("read txt:", timeit(read_txt, number=runs))
print("read pkl:", timeit(read_pkl, number=runs))
print("read npz:", timeit(read_npz, number=runs))

```
